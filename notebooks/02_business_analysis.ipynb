```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    '../data/processed/bank_marketing_clean.csv'
)

df.head()
```

```
df['y'].value_counts()
```

```
df['y'].value_counts(normalize=True) * 100
```

```
conversion_rate = df['y'].mean() * 100

print(
    f'Overall conversion rate: {conversion_rate:.2f}%'
)
```

```
job_conversion = (
    df.groupby('job')['y']
      .agg(['count', 'mean'])
      .sort_values('mean', ascending=False)
)

job_conversion['conversion_rate'] = (
    job_conversion['mean'] * 100
)

job_conversion
```

```
education_conversion = (
    df.groupby('education')['y']
      .agg(['count', 'mean'])
)

education_conversion['conversion_rate'] = (
    education_conversion['mean'] * 100
)

education_conversion.sort_values(
    'conversion_rate',
    ascending=False
)
```

```
housing_conversion = (
    df.groupby('housing')['y']
      .agg(['count', 'mean'])
)

housing_conversion['conversion_rate'] = (
    housing_conversion['mean'] * 100
)

housing_conversion
```
