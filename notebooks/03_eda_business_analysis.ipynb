```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(
    "../data/processed/bank_marketing_clean.csv"
)

df.shape
```

```
conversion_rate = df["y"].mean()

print(f"Customers: {len(df):,}")
print(f"Conversions: {df['y'].sum():,}")
print(f"Conversion rate: {conversion_rate:.2%}")
```

```
df["age"].describe()
```

```
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ]
)
```

```
age_conversion = (
    df.groupby("age_group", observed=True)["y"]
      .agg(
          customers="count",
          conversions="sum",
          conversion_rate="mean"
      )
)

age_conversion["conversion_rate"] *= 100

age_conversion
```

```
job_conversion = (
    df.groupby("job")["y"]
      .agg(
          customers="count",
          conversions="sum",
          conversion_rate="mean"
      )
      .sort_values(
          "conversion_rate",
          ascending=False
      )
)

job_conversion["conversion_rate"] *= 100

job_conversion
```

```
job_conversion["conversion_rate"].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.xlabel("Conversion Rate (%)")
plt.ylabel("Job")
plt.title("Conversion Rate by Job")
plt.tight_layout()
plt.show()
```

```
education_conversion = (
    df.groupby("education")["y"]
      .agg(
          customers="count",
          conversions="sum",
          conversion_rate="mean"
      )
)

education_conversion["conversion_rate"] *= 100

education_conversion.sort_values(
    "conversion_rate",
    ascending=False
)
```

```
df["balance"].describe()
```

```
df["balance_group"] = pd.qcut(
    df["balance"],
    q=5,
    duplicates="drop"
)

balance_conversion = (
    df.groupby("balance_group", observed=True)["y"]
      .agg(
          customers="count",
          conversions="sum",
          conversion_rate="mean"
      )
)

balance_conversion["conversion_rate"] *= 100

balance_conversion
```

```
campaign_conversion = (
    df.groupby("campaign")["y"]
      .agg(
          customers="count",
          conversions="sum",
          conversion_rate="mean"
      )
)

campaign_conversion["conversion_rate"] *= 100

campaign_conversion.head(15)
```

```
campaign_conversion["conversion_rate"].head(15).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.ylabel("Conversion Rate (%)")
plt.xlabel("Number of Contacts")
plt.title("Conversion Rate vs Campaign Contacts")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
