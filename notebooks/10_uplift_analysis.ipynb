# Phase 17 - Uplift Analysis

```
import pandas as pd

df = pd.read_csv("../data/processed/model_data.csv")

df.head()
```

```
campaign_columns = [
    "campaign",
    "previous",
    "pdays",
    "poutcome"
]

df[
    [
        c for c in campaign_columns
        if c in df.columns
    ]
].head()
```

```
df["previous"].describe()
```

```
df["previous"].value_counts()
```

```
df["y"].value_counts(normalize=True)
```

```
df["contact_history"] = (df["previous"] > 0).astype(int)
```

```
group_summary = (
    df
    .groupby("contact_history")["y"]
    .agg(["count", "mean"])
)

group_summary
```

```
import matplotlib.pyplot as plt

rates = df.groupby("contact_history")["y"].mean()

rates.plot(kind="bar")

plt.xlabel("Previous campaign contact")
plt.ylabel("Observed subscription rate")
plt.title("Subscription Rate by Contact History")
plt.xticks([0, 1], ["No previous contact", "Previous contact"], rotation=0)

plt.show()
```

This is **descriptive analysis**, not causal inference.

```
fatigue = (
    df
    .groupby("campaign")["y"]
    .agg(["count", "mean"])
    .reset_index()
)

fatigue.head(20)
```

```
plt.figure(figsize=(8, 5))

plt.plot(fatigue["campaign"], fatigue["mean"], marker="o")

plt.xlabel("Number of contacts during campaign")
plt.ylabel("Observed conversion rate")
plt.title("Campaign Contact Frequency vs Conversion")

plt.show()
```
